```python
print("Types")
types = df.dtypes.to_dict()

print("Type de colonnes")
for column, type in types.items():
    print(f"{column} : {type}")

print()
print()

print("Lignes ayant des valeurs manquantes")
masque_nan = df.isna().any(axis=1)
df_lignes_nan = df[masque_nan]
display(df_lignes_nan)

print()
print()

seuil_quasi_constant = 0.95
quasi_constantes = []
for col in df.columns:
    top_freq = df[col].value_counts(normalize=True, dropna=False).values[0]
    if top_freq >= seuil_quasi_constant:
        quasi_constantes.append((col, round(top_freq * 100, 2)))

print(f"--- Variables quasi-constantes (>= {seuil_quasi_constant*100}%) ---")
for col, pct in quasi_constantes:
    print(f"{col} : {pct}% de valeurs identiques")


seuil_identifiant = 1
identifiants = []
for col in df.columns:
    ratio_unique = df[col].nunique(dropna=False) / len(df)
    if ratio_unique >= seuil_identifiant:
        identifiants.append((col, round(ratio_unique * 100, 2)))

print(f"\n--- Identifiants potentiels (>= {seuil_identifiant*100}% d'unicité) ---")
for col, pct in identifiants:
    print(f"{col} : {pct}% de valeurs uniques")

matrice_corr = df.corr(numeric_only=True).abs()
upper_tri = matrice_corr.where(np.triu(np.ones(matrice_corr.shape), k=1).astype(bool))
seuil_redondance = 0.85

paires_redondantes = []
for col in upper_tri.columns:
    for idx in upper_tri.index:
        val = upper_tri.loc[idx, col]
        if val > seuil_redondance:
            paires_redondantes.append((idx, col, round(val, 2)))

print(f"\n--- Paires de variables redondantes (|r| > {seuil_redondance}) ---")
for v1, v2, r in paires_redondantes:
    print(f"{v1} <--> {v2} (r = {r})")

cols_a_supprimer = list(set(
    [col for col, _ in quasi_constantes] +
    [col for col, _ in identifiants] +
    [v2 for _, v2, _ in paires_redondantes]  # On garde v1 et on retire v2
))

print("\nListe des colonnes à retirer de X :", cols_a_supprimer)
print("\nCependant il est préférable de garder des colonnes comme garage area plutot que garage cars qui sont des variables assez redondantes mais garage_area peut avoir un impact plus imoprtant dans la prédiction de y chapeau")
redondantes_a_retirer = set()
corrs_y = df_encoded.corr(numeric_only=True)['Sale_Price'].abs()
def choisir_meilleure_variable(var1, var2, corrs_y):
    # Règle métier : Donner la priorité aux surfaces (SF, Area)
    mots_cles_surfaces = ['SF', 'Area', 'Total']
    
    var1_est_surface = any(kw in var1 for kw in mots_cles_surfaces)
    var2_est_surface = any(kw in var2 for kw in mots_cles_surfaces)
    
    if var1_est_surface and not var2_est_surface:
        return var2  # On retire var2, on garde var1
    elif var2_est_surface and not var1_est_surface:
        return var1  # On retire var1, on garde var2
    else:
        # Si les deux ou aucune ne sont des surfaces, on prend la meilleure corrélation avec y
        return var1 if corrs_y.get(var1, 0) < corrs_y.get(var2, 0) else var2

for col in upper_tri.columns:
    for row in upper_tri.index:
        if upper_tri.loc[row, col] > 0.85:
            a_retirer = choisir_meilleure_variable(row, col, corrs_y)
            redondantes_a_retirer.add(a_retirer)

print(redondantes_a_retirer)

print("Des variables présentent dans X ont des valeurs quasi constantes, de ce fait le schéma de régression à appliquer est une régression polynomiale dans laquelle les variables avec un poids importants seront valorisées") 
```
